```python
class Company:
    company = "TCS"

    def __init__(self, name, exp):
        self.name = name
        self.exp = exp

    def display(self):
        print(f"Name: {self.name}")
        print(f"Exp: {self.exp}")
        print(f"Company: {self.company}")
        print("----------------")


emp1 = Company("sairam", 5.9)
emp2 = Company("trinadh", 5.6)
emp3 = Company("ramesh", 5.0)

emp1.display()
emp2.display()
emp3.display()

Company.company = "CGI"

emp1.display()
emp2.display()
emp3.display()
```
