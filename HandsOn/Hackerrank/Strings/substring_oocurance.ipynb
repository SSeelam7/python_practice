In this challenge, the user enters a string and a substring. You have to print the number of times that the substring occurs in the given string. String traversal will take place from left to right, not from right to left.

NOTE: String letters are case-sensitive.

Input Format

The first line of input contains the original string. The next line contains the substring.

Constraints


Each character in the string is an ascii character.

Output Format

Output the integer number indicating the total number of occurrences of the substring in the original string.

Sample Input

ABCDCDC
CDC
Sample Output

2

```python
from prompt_toolkit.key_binding.bindings.named_commands import capitalize_word


def count_substring(string, sub_string):
    result = 0

    for i in range((len(string)+1)):
        if string[i:i+len(sub_string)] == sub_string:
            result += 1

    return result

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()


    count = count_substring(string, sub_string)
    print(count)
```

Text alignment using .ljust(width), .rjust(width), center(width)

```python

#thickness = int(input()) #This must be an odd number
thickness = 5 #This must be an odd number

c = 'H'

#Top Cone
for i in range(thickness):
    print((c*i).rjust(thickness-1)+c+(c*i).ljust(thickness-1))

#Top Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))

#Middle Belt
for i in range((thickness+1)//2):
    print((c*thickness*5).center(thickness*6))

#Bottom Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))

#Bottom Cone
for i in range(thickness):
    print(((c*(thickness-i-1)).rjust(thickness)+c+(c*(thickness-i-1)).ljust(thickness)).rjust(thickness*6))
```

Text wrap
You are given a string  and width .
Your task is to wrap the string into a paragraph of width .

string string: a long string
int max_width: the width to wrap to

Returns -- string: a single string with newline characters ('\n') where the breaks should be


ip:
ABCDEFGHIJKLIMNOQRSTUVWXYZ

4

op:
ABCD

EFGH
IJKL

IMNO

QRST

UVWX

YZ

```python
import textwrap

def wrap(string, max_width):
    result = ''
    for itr in range(len(string)):
        if itr % max_width == 0:
            result += (string[itr: itr+max_width]) + '\n'
    return result

if __name__ == '__main__':
    #string, max_width = input(), int(input())
    string, max_width = 'ABCDEFGHIJKLIMNOQRSTUVWXYZ', 4
    result = wrap(string, max_width)
    print(result)
```

```python
msg = 'WELCOME'
length, width = map(int, input().split())
next = 0

#Upper part
for itr in range (1,length+1//2,2):
    print(  ('-' * ((width - (itr*3))//2)).ljust(1)  + ('.|.' * itr) +  ('-' * ((width - (itr*3))//2)).ljust(width) )
    next = itr

#Middle part
print(msg.center(width,'-'))

#Lower part
for itr in range (next,0,-2):
    print(  ('-' * ((width - (itr*3))//2)).ljust(1)  + ('.|.' * itr) +  ('-' * ((width - (itr*3))//2)).ljust(width) )
```

```python
def print_formatted(number):
    width = len(bin(number)[2:])

    for itr in range(1, number + 1):
        print(
            str(itr).rjust(width) + " " +
            oct(itr)[2:].rjust(width) + " " +
            hex(itr)[2:].upper().rjust(width) + " " +
            bin(itr)[2:].rjust(width)
        )

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

```python
string = 'BANANA'
vowels = ['A','E','I','O','U']


c_words = []
v_words = []

for i in range(len(string)+1):
    for j in range(i,len(string)):
        #if string[i] not in vowels and string[i:j+1] not in c_words:
        if string[i].upper() not in vowels:

            c_words.append(string[i:j+1])

        elif string[i].upper() in vowels:
            v_words.append(string[i:j+1])

# print(c_words)
# print(len(c_words))
#
# print(v_words)
# print(len(v_words))

if (len(c_words) == len(v_words)):
    print("Draw")
elif (len(c_words) > len(v_words)):
    print(f"Stuart {len(c_words)}")
else:
    print(f"Kevin {len(v_words)}")


```
